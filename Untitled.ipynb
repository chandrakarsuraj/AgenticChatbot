```python

from langgraph.graph import StateGraph, END, START
from typing import TypedDict, Annotated
from langgraph.graph import MessagesState
from dotenv import load_dotenv
from langchain_core.messages import AIMessage, HumanMessage
from langgraph.checkpoint.sqlite import SqliteSaver
from langchain_openai import ChatOpenAI
import os
import sqlite3

llm  = ChatOpenAI(model = "gpt-6-luna",  openai_api_key=os.environ.get("OPENAI_API_KEY"),)
class State(MessagesState):
    pass

def chat_node(state:State):

    message  = state["messages"]
    return {"messages" :llm.invoke(message)}


conn = sqlite3.connect(database="chatbot.db", check_same_thread=False)
graph = StateGraph(State)
memory = SqliteSaver(conn)

graph.add_node("chat_node", chat_node)

graph.add_edge(START, "chat_node")
graph.add_edge("chat_node", END)
graph = graph.compile(checkpointer=memory)
```

```python
config = {"configurable": {"thread_id": "1"}}

graph.invoke({"messages": [HumanMessage("hellow, how are your")]}, config)
```

```python
import requests
from typing import Optional
from langchain.tools import tool

BASE_URL = "https://api.dharun.dev"


@tool
def cg_mandi_analytics(
    commodity: str,
    district: Optional[str] = None,
    market: Optional[str] = None,
    date: Optional[str] = None,
    limit: int = 100,
):
    """
    Get agricultural mandi prices in Chhattisgarh.

    commodity: Exact commodity name, e.g. Tomato, Onion,
               Paddy(Dhan)(Common), Wheat.
    district: District name, e.g. Raipur, Durg, Mahasamund.
    market: Optional mandi name or partial market name.
    date: Optional date in YYYY-MM-DD format.
    limit: Number of records, between 1 and 1000.
    """

    if not commodity.strip():
        return {"error": "Commodity is required."}

    if not 1 <= limit <= 1000:
        return {"error": "Limit must be between 1 and 1000."}

    params = {
        "commodity": commodity,
        "state": "Chhattisgarh",
        "limit": limit,
    }

    if district:
        params["district"] = district

    if market:
        params["market"] = market

    if date:
        params["date"] = date

    try:
        response = requests.get(
            f"{BASE_URL}/v1/mandi",
            params=params,
            timeout=15,
        )

        response.raise_for_status()
        body = response.json()

        return {
            "commodity": commodity,
            "state": "Chhattisgarh",
            "filters": params,
            "data": body.get("data", []),
        }

    except requests.RequestException as e:
        return {"error": f"Mandi API request failed: {str(e)}"}


@tool
def get_cg_available_commodities():
    """
    Get commodities reported in Chhattisgarh mandis,
    including market counts and modal price ranges.
    """

    try:
        response = requests.get(
            f"{BASE_URL}/v1/mandi/commodities",
            params={"state": "Chhattisgarh"},
            timeout=15,
        )

        response.raise_for_status()
        body = response.json()

        return body.get("data", [])

    except requests.RequestException as e:
        return {"error": f"Commodity API request failed: {str(e)}"}
```

```python
import requests
from langchain.tools import tool

BASE_URL = "https://api.dharun.dev/v1/census"

# Census 2011 state code for Chhattisgarh
CG_STATE_CODE = "22"


@tool
def get_cg_district_population(district: str):
    """
    Get Census 2011 population and demographic information
    for a Chhattisgarh district.

    district: District name, e.g. Raipur, Durg, Bilaspur,
              Mahasamund, Bastar.
    """

    try:
        # Step 1: Search for district
        search_response = requests.get(
            BASE_URL,
            params={
                "q": district,
                "level": "district"
            },
            timeout=10
        )

        search_response.raise_for_status()
        search_body = search_response.json()

        districts = search_body.get("data", [])

        # Keep only Chhattisgarh districts
        cg_districts = [
            item for item in districts
            if item.get("state", {}).get("code") == CG_STATE_CODE
            and item.get("level") == "district"
        ]

        # Step 2: Find exact district name
        matched_district = next(
            (
                item for item in cg_districts
                if item.get("name", "").casefold() == district.casefold()
            ),
            None
        )

        if not matched_district:
            return {
                "error": f"District '{district}' not found in Chhattisgarh.",
                "available_matches": [
                    item.get("name") for item in cg_districts
                ]
            }

        district_code = matched_district["code"]

        # Step 3: Fetch district population
        population_response = requests.get(
            f"{BASE_URL}/district/{district_code}",
            timeout=10
        )

        population_response.raise_for_status()
        population_body = population_response.json()

        return population_body.get("data", {})

    except requests.RequestException as e:
        return {"error": f"Census API error: {str(e)}"}
```

```python
result = get_cg_district_population.invoke({
    "district": "Mahasamund"
})

print(result)
```
