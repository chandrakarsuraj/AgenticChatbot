```python
from langgraph.graph import StateGraph, END, START
from typing import TypedDict, Annotated
from langgraph.graph import MessagesState
from dotenv import load_dotenv
from langchain_core.messages import AIMessage, HumanMessage, AnyMessage
load_dotenv()
```

```python
from langgraph.checkpoint.memory import InMemorySaver
```

```python

from langgraph.graph.message import add_messages


class State(TypedDict):
    messages: Annotated[list[AnyMessage], add_messages]
```

```python
llm.invoke("hello, how are you")
```

```python
from langchain_openai import ChatOpenAI
llm  = ChatOpenAI(model = "gpt-6-luna")
graph = StateGraph(State)

class State(MessagesState):
    pass

def chat_node(state:State):

    message  = state["messages"]
    return {"messages" :llm.invoke(message)}

memory = InMemorySaver()

graph.add_node("chat_node", chat_node)

graph.add_edge(START, "chat_node")
graph.add_edge("chat_node", END)
graph = graph.compile(checkpointer=memory)
```

```python
config = {"configurable": {"thread_id": "1"}}
res = graph.stream({"messages": [HumanMessage("hellow, how are your")]},  config, stream_mode="messages",version="v2")
```

```python
from rich import print
for chunk in res:
    print(chunk)
```

```python
config = {"configurable": {"thread_id": "1"}}
while True:
    user_message = input("Type here ..")
    print("User :", user_message)
    if user_message == "exit":
        break
    res = graph.invoke({"messages": [HumanMessage(user_message)]}, config)
    print("AI message: " , res["messages"][-1].content)
    
    
```

```python
from rich import print
print(graph.get_state(config))
```

```python
from langgraph.graph.message import add_messages
import time

class newState(TypedDict):
    messages: Annotated[list[AnyMessage], add_messages]
    messages1: Annotated[list[AnyMessage], add_messages]
    messages2: Annotated[list[AnyMessage], add_messages]
    messages3: Annotated[list[AnyMessage], add_messages]


def step1(newState):
    return {"messages1": "Hellow from node1"}


def step2(newState):
    time.sleep(30)
    return {"messages2": "Hellow from node2"}

def step3(newState):
    return {"messages2": "Hellow from node3"}

memory = InMemorySaver()

graph = StateGraph(newState)
graph.add_node("node1", step1)
graph.add_node("node2", step2)
graph.add_node("node3", step3)



graph.add_edge(START, "node1")
graph.add_edge("node1", "node2")
graph.add_edge("node2", "node3")


graph.add_edge("node3", END)
graph = graph.compile(checkpointer=memory)
```
